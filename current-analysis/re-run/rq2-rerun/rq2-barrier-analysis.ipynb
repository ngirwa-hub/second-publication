```python
# cell3: merging zeroshot and context data of barriers seelect responses
import pandas as pd

# load data
df_ctx = pd.read_csv('barriers-ctx-zeroshot-working/contextBarrierSelectResponses_50.csv')
df_zeroshot = pd.read_csv('../1_zero-shot/zeroshotResponses/zeroshotBarrierSelectResponses.csv')

# define source and condition
df_ctx['condition'] = 'context'
df_zeroshot['condition'] = 'zeroshot'

# exact columns to keep
columns_to_keep = ['row_id', 'variant_id', 'base_model', 'model', 'barrier_id', 'official_label', 'condition', 'iteration']

# combine the dataframes
df_combined = pd.concat([df_ctx[columns_to_keep], df_zeroshot[columns_to_keep]], ignore_index=True)

# save combined dataframe
df_combined.to_csv(output_dir / 'barriers-ctx-zeroshot-responses.csv', index=False)
```

```python
# jsi latex table
from pathlib import Path
import pandas as pd
from IPython.display import display, Markdown

# Input generated by barrier-select-rq2.R
input_path = Path(
    "barriers-ctx-zeroshot-working/top5_overlap_summary_by_family.csv"
)

results = pd.read_csv(input_path)

# Labels and ordering used in the paper
model_labels = {
    "gemma3": "Gemma3:12B",
    "llama": "Llama-Pro",
    "mistral": "Mistral",
    "phi4": "Phi4",
}

model_order = [
    "Phi4",
    "Gemma3:12B",
    "Mistral",
    "Llama-Pro"
]

jaccard_table = (
    results[["family", "jaccard"]]
    .assign(Model=lambda x: x["family"].map(model_labels))
    .rename(columns={"jaccard": "JSI"})
    [["Model", "JSI"]]
    .set_index("Model")
    .reindex(model_order)
    .reset_index()
)

latex_header = (
    r"$JSI(\mathrm{Context},\mathrm{Zero\text{-}shot})$"
)

jaccard_table = jaccard_table.rename(
    columns={"JSI": latex_header}
)

latex_table = jaccard_table.to_latex(
    index=False,
    escape=False,
    column_format="lr",
    caption="JSI on top-5 barriers for context vs. zero-shot",
    label="tab:jaccard-top5-barriers",
    float_format=lambda value: f"{value:.3f}",
    position="htbp",
)
# save the LaTeX table to a file
output_path = Path("barriers-ctx-zeroshot-working/jaccard_top5_barriers.tex")
output_path.write_text(latex_table)
print(latex_table)
```
