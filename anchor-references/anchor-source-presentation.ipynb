```python
def latex_escape(text):
    if pd.isna(text):
        return ""
    text = str(text)
    replacements = {
        "&": r"\&",
        "%": r"\%",
        "$": r"\$",
        "#": r"\#",
        "_": r"\_",
        "{": r"\{",
        "}": r"\}",
    }
    for old, new in replacements.items():
        text = text.replace(old, new)
    return text

def make_multirow_latex_table(df_visible, caption, label):
    grouped = []
    for anchor_type, group in df_visible.groupby(["Anchor type", "Anchor phrase"], sort=False):
        grouped.append((anchor_type[0], anchor_type[1], group.reset_index(drop=True)))

    lines = []
    lines.append(r"\begin{longtable}{p{1.8cm} p{5.0cm} p{2.2cm} p{0.9cm} p{2.6cm} p{3.0cm}}")
    lines.append(r"\caption{" + caption + r"}\label{" + label + r"} \\")
    lines.append(r"\toprule")
    lines.append(
        r"\textbf{Anchor type} & \textbf{Anchor phrase} & \textbf{Document type} & "
        r"\textbf{Year} & \textbf{Author/Org.} & \textbf{Citation} \\"
    )
    lines.append(r"\midrule")
    lines.append(r"\endfirsthead")

    lines.append(r"\toprule")
    lines.append(
        r"\textbf{Anchor type} & \textbf{Anchor phrase} & \textbf{Document type} & "
        r"\textbf{Year} & \textbf{Author/Org.} & \textbf{Citation} \\"
    )
    lines.append(r"\midrule")
    lines.append(r"\endhead")

    lines.append(r"\midrule")
    lines.append(r"\multicolumn{6}{r}{\textit{Continued on next page}} \\")
    lines.append(r"\midrule")
    lines.append(r"\endfoot")

    lines.append(r"\bottomrule")
    lines.append(r"\endlastfoot")

    for anchor_type, anchor_phrase, group in grouped:
        n = len(group)
        for i, (_, row) in enumerate(group.iterrows()):
            cells = []
            if i == 0:
                if n == 1:
                    cells.append(latex_escape(anchor_type))
                    cells.append(latex_escape(anchor_phrase))
                else:
                    cells.append(rf"\multirow{{{n}}}{{*}}{{{latex_escape(anchor_type)}}}")
                    cells.append(rf"\multirow{{{n}}}{{5.0cm}}{{{latex_escape(anchor_phrase)}}}")
            else:
                cells.append("")
                cells.append("")

            cells.extend([
                latex_escape(row["Document type"]),
                latex_escape(row["Year"]),
                latex_escape(row["Author/Organization"]),
                latex_escape(row["Citation"]),
            ])

            lines.append(" & ".join(cells) + r" \\")
        lines.append(r"\midrule")

    if lines[-1] == r"\midrule":
        lines.pop()

    lines.append(r"\end{longtable}")
    return "\n".join(lines)
```

```python
visible_cols = [
    "Anchor type",
    "Anchor phrase",
    "Document type",
    "Year",
    "Author/Organization",
    "Citation",
]

latex_table = make_multirow_latex_table(
    df[visible_cols],
    caption="Sources used to support each anchor phrase in the bias framing prompts.",
    label="tab:anchor_sources",
)

print(latex_table)

# Save LaTeX to file
latex_path = out_dir / "anchor_sources_table_new.tex"
latex_path.write_text(latex_table, encoding="utf-8")
latex_path
```
